```python
# Import necessary libraries
from pyspark.sql import SparkSession
from pyspark.sql.functions import col
```

```python
# Create Spark session
spark = SparkSession.builder.appName("Parking").getOrCreate()
```

```python
spark.conf.set("fs.azure.account.key.<blob-storgae-account-name>.blob.core.windows.net", "<access-keys>")
```

```python
file_location = "wasbs://<blob-container-name>@<blob-storage-account-name>.blob.core.windows.net"
```

```python
# Read CSV into DataFrame

df_bronze = spark.read.format("csv") \
    .option("header", "true") \
    .option("inferSchema", "true") \
    .load(file_location)
```

## Check for Trailing Spaces in the Column

```python
for col_name in df_bronze.columns:
    if col_name != col_name.strip():
        print(f"Column with trailing/leading spaces: '{col_name}'")
```

## Rename column to remove trailing spaces

```python
df_bronze = df_bronze.withColumnRenamed("Days Parking In Effect    ", "Days Parking In Effect")
```

```python
df_mine = df_bronze
```

```python
import re

def clean_column_name(name):
    name = name.strip()  # Remove leading/trailing spaces
    name = re.sub(r'[ ,;{}()\n\t=?:]', '_', name)  # Replace unsupported characters with underscore
    name = re.sub(r'_+', '_', name)               # Collapse multiple underscores
    return name.strip('_')                        # Remove leading/trailing underscores

# Rename columns in DataFrame
df_bronze = df_bronze.toDF(*[clean_column_name(c) for c in df_bronze.columns])
```

```python
df_bronze.write.format("delta").mode("overwrite").save(bronze_path)
```
