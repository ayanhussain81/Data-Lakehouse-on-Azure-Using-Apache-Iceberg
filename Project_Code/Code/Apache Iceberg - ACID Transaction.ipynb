### Atomic Update (Isolated Write)

```python
from pyspark.sql.functions import when, col

# Load original table
df = spark.table("iceberg_catalog.iceberg_db.parking_iceberg")
```

```python
# Update the Vehicle_Color where condition matches
updated_df = df.withColumn(
    "Vehicle_Color",
    when(col("Summons_Number") == 1326051398, "GOLD").otherwise(col("Vehicle_Color"))
)
```

```python
updated_df.writeTo("iceberg_catalog.iceberg_db.parking_iceberg").overwritePartitions()
```

```python
%sql
-- Verify the update

SELECT Vehicle_Color 
FROM iceberg_catalog.iceberg_db.parking_iceberg 
WHERE Summons_Number = 1326051398; -- Should return 'GOLD'
```

### Concurrent Read During Write

#### Rollback Test

```python
from pyspark.sql import SparkSession

spark = SparkSession.builder.getOrCreate()

# Table you're writing to
table_name = "iceberg_catalog.iceberg_db.parking_iceberg"

# Step 1: Capture the current snapshot ID BEFORE any operation
snapshot_id_df = spark.sql(f"""
    SELECT snapshot_id 
    FROM {table_name}.snapshots 
    ORDER BY committed_at DESC 
    LIMIT 1
""")
snapshot_id = snapshot_id_df.first()[0]
print(f"Captured snapshot before write: {snapshot_id}")

try:
    # Step 2: Your data write (append mode)
    new_df = spark.createDataFrame([
        (12345678, "ABC123", "NY", "PAS", "2023-10-01", 36, "SEDAN", "HONDA", "NYPD", 202510, 100, 20, 101, "MAN", "TEST_ACID")
    ], ["Summons_Number", "Plate_ID", "Registration_State", "Plate_Type", "Issue_Date", "Violation_Code", "Vehicle_Body_Type", "Vehicle_Make", "Issuing_Agency", "Vehicle_Expiration_Date", "Violation_Location", "Violation_Precinct", "Issuer_Precinct", "Violation_County", "Violation_Description"])

    # Step 3: Perform the write
    new_df.writeTo(table_name).append()
    print("Data appended successfully.")

    # Step 4: Simulate failure (optional - for testing rollback)
    raise Exception("Simulated failure after append!")

except Exception as e:
    print(f" Error occurred: {e}")
    print(f"Rolling back to snapshot: {snapshot_id}")

    # Step 5: Read from the safe snapshot
    snapshot_df = spark.read.format("iceberg") \
        .option("snapshot-id", snapshot_id) \
        .table(table_name)

    # Step 6: Overwrite the current state with that snapshot (rollback)
    snapshot_df.writeTo(table_name).overwritePartitions()
    print("Rollback complete — table restored to pre-write state.")
```
